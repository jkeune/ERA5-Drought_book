## Options to download ERA5-Drought data

ERA5-Drought data is currently hosted on the Climate Data Store (CDS, [CDS User Guide](https://confluence.ecmwf.int/display/CKB/Climate+Data+Store+%28CDS%29+documentation))).
There are two options to download data from the CDS:
1. **CDS web interface**: through manual selections via the web interface, accessible [here](https://cds.climate.copernicus.eu/datasets/derived-drought-historical-monthly?tab=download).
2. **CDS Application Programming interface**: through a dedicated CDS Application Programming Interface (API). Instruction on how to set this up, see [here](https://cds.climate.copernicus.eu/how-to-api). 

For either option, you need to be a registered user in the CDS — please register through the CDS homepage. 

Using the web interface, you can manually select the drought indices, the time span, and the region and the submit the request to the system. The result is a downloadable netCDF or zip file. Just click on [`Downloads`]() on the ERA5-Drought site and click through the sections. You have to accept the license before you can submit the request.

Using the CDS API, you can download the data directly from within python. 
Please note: you can use the CDS web interface to construct the API request for python. Just open the web interface, click your way through the sections, and then click the 'Show API request' button to get the code that you need to copy into python. One example of such an API request is:

### Download ERA5-Drought using the CDS API

```python
import cdsapi
 
dataset = "derived-drought-historical-monthly"
request = {
    "variable": ["standardised_precipitation_index"],
    "accumulation_period": ["12"],
    "version": "1_0",
    "product_type": ["reanalysis"],
    "dataset_type": "consolidated_dataset",
    "year": ["2023"],
    "month": [
        "01", "02", "03",
        "04", "05", "06",
        "07", "08", "09",
        "10", "11", "12"
    ]
}
 
client = cdsapi.Client()
client.retrieve(dataset, request).download()
```

This code downloads the global SPI-12 for all months in 2023 using the consolidated ERA5 data set. 

### Download a subset of ERA5-Drought

If you do not require global data, but only data for, e.g., Mozambique, you can set the extent of the data to be downloaded. The code below downloads the SPI-12 as above but only for the extent of Mozambique using the key `area` ([North, West, South, East]).

```python
import cdsapi

dataset = "derived-drought-historical-monthly"
request = {
    "variable": ["standardised_precipitation_index"],
    "accumulation_period": [
        "1",
        "12"
    ],
    "version": "1_0",
    "product_type": ["reanalysis"],
    "dataset_type": "consolidated_dataset",
    "year": ["2023"],
    "month": [
        "01", "02", "03",
        "04", "05", "06",
        "07", "08", "09",
        "10", "11", "12"
    ],
    "area": [-10, 30, -27, 41]
}

client = cdsapi.Client()
client.retrieve(dataset, request).download()
```
